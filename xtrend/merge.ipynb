```python
import pandas as pd

data_1 =  pd.read_csv("pinnacle_features.csv")
data_1.drop(columns=["date"], inplace=True)
```

```python
import os 
dataset = []
for dirpath, dirnames, filenames in os.walk("dataset/CPD/pinnacle_cpd_63lbw/"):
    for filename in filenames:
        if filename.endswith(".csv"):
            filepath = os.path.join(dirpath, filename)
            ticker = filename.split(".")[0]
            data_2 = data_1[data_1["ticker"]==ticker]
            data_3 = pd.read_csv(filepath)
            data_4 = pd.merge(data_3, data_2, left_on="date", right_on="Date", how="left")
            dataset.append(data_4)
```

```python
pd.concat(dataset).rename(columns={"cp_location_norm":"cp_rl_63", "cp_score":"cp_score_63"}).to_csv("final_pinnacle_cpd_63lbw.csv", index=False)
```
