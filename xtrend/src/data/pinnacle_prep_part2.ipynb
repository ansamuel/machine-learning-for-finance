```python
import pandas as pd
import numpy as np
import torch

BATCH_SIZE = 16
NUM_CONTEXT = 3
```

```python
train_data_prepped_all_segments = pd.read_pickle("prepped.pkl")
```

```python
train_data_prepped_all_segments["date"].max() # needs fixing
```

```python
dfs = []
items_per_batch = (NUM_CONTEXT + 1) * BATCH_SIZE
for i, group in train_data_prepped_all_segments.groupby(["ticker", "seq_len"]):
    num_rows = (len(group) // items_per_batch) * items_per_batch
    if not num_rows:
        next
    # print(num_rows)
    # print(len(group))
    dfs.append(group.sample(frac=1)[:num_rows])
    # print(group.sample(frac=1)[:num_rows]["seq_len"])
df = pd.concat(dfs).reset_index()
```

```python
len(df)/len(train_data_prepped_all_segments)
```

```python
shuffled = (
    df
    .assign(group_var=df.index // items_per_batch)
    .set_index("group_var")
    .loc[np.random.permutation(len(df) // items_per_batch)]
)
```

```python
shuffled.reset_index()["group_var"].head(66)
```

```python
for start in range(0, len(shuffled), items_per_batch):
    batch = shuffled.iloc[start : start + items_per_batch]
    seq_len = batch["seq_len"].iloc[0]
    x = torch.reshape(
        torch.cat(batch["x"].tolist()).unsqueeze(0),
        [NUM_CONTEXT + 1, BATCH_SIZE, seq_len, 1],
    ).permute(1, 2, 3, 0)

    x_context = x[:,:,:,:NUM_CONTEXT]
    x_test = x[:,:,:,-1:]

    y = torch.reshape(
        torch.cat(batch["y"].tolist()).unsqueeze(0),
        [NUM_CONTEXT + 1, BATCH_SIZE, seq_len, 1],
    ).permute(1, 2, 3, 0)

    y_context = y[:,:,:,:NUM_CONTEXT]
    y_test = y[:,:,:,-1:]

    # if batch["x"].iloc[0].shape[1] == 3:
    #     break
```
